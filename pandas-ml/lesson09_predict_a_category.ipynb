# Lesson 9: Predict a category

**Introduction to Machine Learning with pandas** (lesson 9 of 12)

By the end you can train a decision tree to predict a category, and read its accuracy.

The first two cells open the clean table, `penguins_clean.csv`.

```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins_clean.csv")
```

## 1. Choose the features and the target

The target is now a category: the species. Body mass was the target in lesson 8 and is a feature here, because a column is a feature or a target depending on the question you ask. Bill depth and body mass are the pair the tree on the slide uses; they separate the species fairly well.

```python
features = ["bill_depth_mm", "body_mass_g"]
X = penguins[features]
y = penguins["species"]
```

## 2. Split the rows

The same split as in lessons 7 and 8.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

## 3. Make the tree

`DecisionTreeClassifier` is a model that asks yes-or-no questions about the features. `random_state=42` gives the same tree on every run.

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(random_state=42)
```

## 4. Fit and predict

The same two steps as in lesson 8. The predictions are now species names rather than grams: that is what predicting a category looks like. `predicted` is kept for lesson 10, which compares it with the true species.

```python
model.fit(X_train, y_train)
predicted = model.predict(X_test)
predicted[:5]
```

## 5. Read the accuracy

**Pause and reflect:** before you run it, guess the share of test penguins the tree will name correctly. Would you be happy with half right?

`score` takes the test features and the true species, does the predicting itself, and gives the share of test rows it got right. 0.72 means about 62 of the 86 test penguins were named correctly.

```python
accuracy = model.score(X_test, y_test)
round(accuracy, 2)
```

## Task: predict a category in your own data

1. Open `my_data.ipynb`.
2. Add the cell below. Change `"penguins_clean.csv"` to your file, the list to number columns of your own, and the target to a text column.
3. Report the accuracy. Half right would be 0.5; the next lesson shows how to judge the score.

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
my_data = pd.read_csv("penguins_clean.csv")
features = ["bill_length_mm", "flipper_length_mm"]
X = my_data[features]
y = my_data["island"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)
accuracy = model.score(X_test, y_test)
round(accuracy, 2)
```
