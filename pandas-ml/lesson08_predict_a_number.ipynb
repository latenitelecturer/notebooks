The first two cells open the clean table, `penguins_clean.csv`.

```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins_clean.csv")
```

## 1. Prepare the data

These are the lines from lesson 7: the features, the target, and the split. Run this cell before the others, or the names `X_train` and `y_train` will not exist.

```python
from sklearn.model_selection import train_test_split
features = ["flipper_length_mm"]
X = penguins[features]
y = penguins["body_mass_g"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

## 2. Make the model

`LinearRegression` is a model that draws a straight line through the points. Linear means "in a straight line". The model is empty until it is fitted.

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

## 3. Fit the model

`fit` shows the model the training rows. `coef_` is short for coefficient, which is the slope of the line: the grams of body mass added by each extra millimetre of flipper. The underscore at the end marks a value the model learned by fitting. The square brackets in the output mean one number per feature.

```python
model.fit(X_train, y_train)
slope = model.coef_
slope
```

## 4. Predict

`predict` reads a body mass off the line for each test row. It gives back a plain list of numbers, not a pandas table, so `[:5]` shows the first five instead of `head()`. They are rounded to whole grams.

```python
predicted = model.predict(X_test)
rounded = predicted.round()
rounded[:5]
```

## 5. Measure the error

The mean absolute error is the average distance between a prediction and the true value. It is in grams, like the target.

**Pause and reflect:** before you run it, guess the error. The penguins weigh between 2,700 and 6,300 g.

```python
from sklearn.metrics import mean_absolute_error
error = mean_absolute_error(y_test, predicted)
round(error)
```

## 6. Compare with a simple guess

Is that error good? Compare it with the simplest possible prediction: always guess the average body mass of the training rows. The third line makes a list with the average repeated once for every test row. Compare the two errors: the smaller one is the better prediction.

```python
average_mass = y_train.mean()
n_test = len(y_test)
always_average = [average_mass] * n_test
guess_error = mean_absolute_error(y_test, always_average)
round(guess_error)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
my_data = pd.read_csv("penguins_clean.csv")
features = ["bill_length_mm", "bill_depth_mm"]
X = my_data[features]
y = my_data["body_mass_g"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
model = LinearRegression()
model.fit(X_train, y_train)
predicted = model.predict(X_test)
error = mean_absolute_error(y_test, predicted)
round(error)
```
