```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins_clean.csv")
```

## 1. Draw a histogram

A histogram shows how the values in one column are spread out. Each bar counts the penguins in one range of body mass.

Storing the chart under the name `chart` stops the notebook printing a line of text above it; the chart appears either way. On the chart, the bottom axis is body mass in grams, and `Frequency` up the side is the number of penguins in each bar.

```python
body_mass = penguins["body_mass_g"]
chart = body_mass.plot.hist()
```

## 2. Change the number of bars

`bins=20` is a named argument: a name, an equals sign, and a value. It asks for twenty bars.

**Pause and reflect:** will more bars make the shape smoother, or more jagged?

```python
chart = body_mass.plot.hist(bins=20)
```

## 3. Draw a scatter plot

A scatter plot compares two columns. Each dot is one penguin. `x` names the column along the bottom, and `y` the column up the side.

This chart is one colour. Colouring the dots by species needs another argument, which lesson 12 uses.

```python
chart = penguins.plot.scatter(x="flipper_length_mm", y="body_mass_g")
```
