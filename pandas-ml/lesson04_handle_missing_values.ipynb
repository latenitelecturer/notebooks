The first two cells open the data, as before.

```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins.csv")
```

## 1. Count the missing values

`isna` marks every missing value as `True` and every other value as `False`. `sum` counts each `True` as 1, so it gives the number of gaps in each column.

**Pause and reflect:** before you run it, guess which column has the most gaps.

```python
is_missing = penguins.isna()
is_missing.sum()
```

## 2. Look at the rows with a gap

`isna` on one column gives a `True` or `False` answer for every row. Putting those answers inside the square brackets keeps only the `True` rows, as in lesson 3. These are the penguins with no bill length recorded. pandas shows a missing value as `NaN`, which stands for "not a number".

```python
bill_length = penguins["bill_length_mm"]
no_bill_length = bill_length.isna()
penguins[no_bill_length]
```

## 3. Drop the rows with no measurements

`dropna` removes rows with a gap. `subset=measurements` is a **named argument**: the name of a setting, an equals sign, and its value. It names the columns to check, so a row is removed only when a measurement is missing.

```python
measurements = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
penguins_clean = penguins.dropna(subset=measurements)
penguins_clean.shape
```

## 4. Fill the gaps in a text column

`fillna` replaces each gap with a value you choose. The third line stores the filled column back in the table, so the change is kept. Rows 8 to 11 had no sex recorded; they now say `unknown`.

```python
sex = penguins_clean["sex"]
sex_filled = sex.fillna("unknown")
penguins_clean["sex"] = sex_filled
penguins_clean.head(12)
```

## 5. Save the clean table

Later lessons open this file. `index=False` is another named argument: it leaves the row numbers out of the file.

```python
penguins_clean.to_csv("penguins_clean.csv", index=False)
```
