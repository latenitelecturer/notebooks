The first two cells open the clean table, `penguins_clean.csv`.

```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins_clean.csv")
```

## 1. Choose the features

The features are the columns the model may look at. They are written as a list of column names, even when there is only one, because the table of features must be a table. It is called `X`, with a capital letter, because it is a table.

```python
features = ["flipper_length_mm"]
X = penguins[features]
X.head()
```

## 2. Choose the target

The target is the column the model will predict. It is called `y`, with a small letter, because it is a single column.

```python
y = penguins["body_mass_g"]
y.head()
```

## 3. Split the rows

`train_test_split` picks one row in every four at random and keeps them back as the test set; the rest form the training set. `random_state=42` fixes which rows are picked, so the split is the same on every run.

The four names come back in this fixed order: training features, test features, training target, test target. The library is installed as `scikit-learn` but imported as `sklearn`; both are the same thing.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

## 4. Check the shapes

**Pause and reflect:** before you run it, work out what the four shapes should be. Three rows in every four went to the training set, out of 342.

`(256, 1)` means 256 rows and 1 column. `(256,)` means 256 values with no column count, because `y` is a single column. The training set has 256 rows and the test set 86. Together they make 342.

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```
