```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins.csv")
```

## 1. Select one column

Square brackets with a column name select that column. The result is one column with its 344 values. The `Name`, `Length` and `dtype` printed under it can be ignored.

```python
penguins["species"]
```

## 2. Ask a question of every row

One equals sign stores a value. Two equals signs ask a question: is the species Gentoo? The answer is `True` or `False` for each row, and the answers are stored under the name `is_gentoo`.

The second line shows rows 150 to 154: the last two Adelie penguins and the first three Gentoo.

```python
is_gentoo = penguins["species"] == "Gentoo"
is_gentoo[150:155]
```

## 3. Keep the rows that match

Square brackets with the True and False answers inside keep the rows where the answer is `True`. This is called a **filter**. `shape` shows how many rows were kept.

```python
gentoo = penguins[is_gentoo]
gentoo.shape
```

## 4. Compare numbers

`>` means greater than. This keeps the penguins heavier than 5000 grams.

**Pause and reflect:** will there be more or fewer than the 124 Gentoo?

```python
is_heavy = penguins["body_mass_g"] > 5000
heavy = penguins[is_heavy]
heavy.shape
```

## 5. The original table is unchanged

A filter makes a new table under a new name. `penguins` still holds every row.

```python
penguins.shape
```

## 6. Sort the rows

`sort_values` orders the rows by one column, smallest first. `ascending=False` is an option, written after a comma, that puts the largest value first.

```python
by_mass = penguins.sort_values("body_mass_g", ascending=False)
by_mass.head()
```
