The first two cells open the clean table, `penguins_clean.csv`.

```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins_clean.csv")
```

## 1. Train the tree from lesson 9

The same features, target, split and model, in one cell. Run it first. Its accuracy on the test rows is 0.72.

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
features = ["bill_depth_mm", "body_mass_g"]
X = penguins[features]
y = penguins["species"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)
test_score = model.score(X_test, y_test)
round(test_score, 2)
```

## 2. Work out the baseline

The baseline is the score of a very simple guess: always answer with the most common species in the training rows. The first cell counts the species; read its top line.

The second cell asks, for every test penguin, "is it an Adelie?". Python counts each `True` as 1 and each `False` as 0, so the mean of those answers is the share of test penguins that are Adelie: the score the simple guess would get.

```python
counts = y_train.value_counts()
counts
```

```python
is_adelie = y_test == "Adelie"
baseline = is_adelie.mean()
print(baseline)
```

## 3. Compare the training score with the test score

A model that scores far higher on its training rows than on the test rows has memorised them. This is called **overfitting**.

**Pause and reflect:** before you run it, guess the training score. Will it be higher or lower than the test score from section 1?

```python
train_score = model.score(X_train, y_train)
print(train_score, test_score)
```

## 4. Limit the tree

`max_depth=3` allows only three questions in a row, so the tree cannot memorise every training row.

```python
small_tree = DecisionTreeClassifier(max_depth=3, random_state=42)
small_tree.fit(X_train, y_train)
small_train = small_tree.score(X_train, y_train)
small_test = small_tree.score(X_test, y_test)
print(small_train, small_test)
```

## 5. Build the confusion matrix

`confusion_matrix` counts each true species against each predicted species. Each row is a true species, and each column is a predicted species. `classes_` is the list of species names the tree learned, in the order the matrix uses; the `pd.DataFrame` line only adds those names as labels, so the table is easier to read.

The correct answers lie on the diagonal: 42, 0 and 25. The Chinstrap row has no correct answers.

```python
from sklearn.metrics import confusion_matrix
predicted = small_tree.predict(X_test)
matrix = confusion_matrix(y_test, predicted)
species_names = small_tree.classes_
table = pd.DataFrame(matrix, index=species_names, columns=species_names)
table
```
