## 1. Import pandas

pandas is a **library**: a set of ready-made tools that other people wrote, which you can use in your own code. `import` loads it, and `as pd` gives it a short name to type.

```python
import pandas as pd
```

## 2. Read the file

`pd.read_csv` opens a CSV file and returns a table. The `=` stores that table under the name `penguins`. A name that holds a value is called a **variable**; you use the name whenever you want the value again.

Nothing is printed when you run this cell, because the table has been stored, not shown.

```python
penguins = pd.read_csv("penguins.csv")
```

## 3. Look at the first rows

`head` shows the first five rows. Each row is one penguin. Each column is one thing that was recorded about it.

A notebook shows the value of the last line of a cell, which is why `head()` appears without `print`. The numbers down the left-hand side count the rows, starting from 0. `NaN` means a value is missing; lesson 4 deals with that.

```python
penguins.head()
```

## 4. Check the size of the table

`shape` gives two numbers: rows first, then columns.

**Pause and reflect:** before you run the cell, guess the two numbers. The cell above showed rows 0 to 4 and six of the columns.

```python
penguins.shape
```
