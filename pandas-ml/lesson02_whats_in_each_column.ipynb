```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins.csv")
```

## 1. List the columns

`columns` gives the name of every column, in order. Read the names inside the square brackets; the `Index(` at the start and the `dtype='object'` at the end can be ignored for now.

```python
penguins.columns
```

## 2. See the type of each column

`dtypes` gives the **data type** of each column: the kind of value it holds. `float64` is a number with a decimal point, `int64` is a whole number, and `str` is text (short for string). The 64 can be ignored.

If your copy of pandas is older, text columns show `object` instead of `str`. Both mean text.

```python
penguins.dtypes
```

## 3. Summarise the number columns

`describe` reports, for each number column, how many values are present (`count`), the average (`mean`), and the smallest and largest values (`min`, `max`). The other rows (`std`, `25%`, `50%`, `75%`) can be ignored until later.

**Pause and reflect:** before you run it, say which columns will appear. `describe` reports on number columns only.

```python
penguins.describe()
```
