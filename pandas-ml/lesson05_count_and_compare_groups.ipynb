```python
import pandas as pd
```

```python
penguins = pd.read_csv("penguins_clean.csv")
```

## 1. Count the categories

`value_counts` counts how many rows have each value. The most common value comes first.

```python
species = penguins["species"]
species.value_counts()
```

## 2. Compare the average of each group

`groupby` splits the rows into groups, one for each species, and stores them under the name `by_species`. On its own, `by_species` shows nothing useful: you must choose a column to summarise, and a summary. The second line chooses body mass. `mean` then gives one average for each group.

The long decimals are normal. Adelie and Chinstrap are both about 3,700 g, and Gentoo about 5,100 g.

```python
by_species = penguins.groupby("species")
mass_by_species = by_species["body_mass_g"]
mass_by_species.mean()
```

## 3. Summarise another column

The same groups can summarise any number column.

**Pause and reflect:** which species has the longest flippers?

```python
flipper_by_species = by_species["flipper_length_mm"]
flipper_by_species.mean()
```
